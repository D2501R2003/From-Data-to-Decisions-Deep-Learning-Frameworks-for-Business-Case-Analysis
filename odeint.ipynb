```python
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt
```

```python
#function that returns dy/dt
def model(y,t):
  k = 0.3
  dydt = -k*y
  return dydt
```

```python
# initial condition
y0 = 5

# time point
t = np.linspace(0,20, 50)

```

```python
# solve ODE
y = odeint(model, y0, t)
```

```python
# plot results
plt.plot(t,y)
plt.xlabel('time')
plt.ylabel('y(t)')
plt.show()
```

```python
# function that returns dy/dt
def model(y,t,k):
  dydt = -k * y
  return dydt
```

```python
# solve ODEs
k = 0.1
y1 = odeint(model, y0, t, args=(k,))
k = 0.2
y2 = odeint(model, y0, t, args=(k,))
k = 0.5
y3 = odeint(model, y0, t, args=(k,))
```

```python
# plot results
plt.plot(t,y1, 'r-', linewidth=2, label='k=0.1')
plt.plot(t,y2, 'b--', linewidth=2, label='k=0.2')
plt.plot(t,y3, 'g:', linewidth=2, label='k=0.5')
plt.xlabel('time')
plt.ylabel('y(t)')
plt.legend(loc = 'best')
plt.show()
```

```python
def model(y,t):
  dydt = -y + 1.0
  return dydt

# initial condition
y0 = 0
```

```python
#time points
t = np.linspace(0,10, 100)

# solve ODE
y = odeint(model, y0, t)
```
