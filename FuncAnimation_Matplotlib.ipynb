### FuncAnimation in Python
matplotlib.animation.FuncAnimation(fig, func, frames=None, init_func=None, fargs=None, interval)

Relevant inputs:

*   fig: The figure we want to animate
*   func: A function that updates our data(func(frame, *fargs))
*   frames: Total number of frames
*   init_func: Initial plot
*   fargs: Additional Arguments
*   interval: Delay between each frames in (ms)

(Optional)
*  blit-> Bool: Optimizing drawing

Methods:
*   pause(): Pause the animation
*   resume(): Resume the animation
*   Save(): Saves the animation.













```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import numpy as np
```

```python
# Sine Wave animation
t = np.linspace(0, 10,100)
y = np.sin(t)
```

```python
fig, axis = plt.subplots()

axis.plot(t,y)

plt.show()
```

```python
# Using FuncAnimation
fig, axis = plt.subplots()

axis.set_xlim(min(t), max(t))
axis.set_ylim([-2,2])
animated_plot, = axis.plot([], [])

def update_data(frame):

      animated_plot.set_data(t[:frame],y[:frame])
      return animated_plot,


animation = FuncAnimation(
                fig = fig,
                func = update_data,
                frames = len(t),
                interval = 25,


)
animation.save("animation_sin.gif")
from IPython.display import HTML

HTML(animation.to_jshtml())





```
