```python
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt
```

```python
def diff_eqn(y, x):
    dydx = x
    return dydx
```

```python
x = np.linspace(0,5,10)
```

```python
y = odeint(diff_eqn,0,x)
```

```python
plt.xlabel('x')
plt.ylabel('y')
plt.plot(x,y)
plt.grid()
plt.legend('y')
```
